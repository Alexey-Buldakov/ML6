# Задание 1. Ответы на вопросы из Preamble и Introduction

В этом ноутбуке приведены теоретические ответы на вопросы, требуемые в проекте по снижению размерности.

## 1. Для каких ранее изученных моделей актуальна «проклятие размерности»?

«Проклятие размерности» особенно актуально для алгоритмов, которые используют расстояния между объектами, например **K-ближайших соседей (KNN)** и **K-means**.

При увеличении числа признаков расстояния между точками становятся менее информативными: объекты начинают выглядеть более похожими с точки зрения расстояния, а для сохранения той же плотности наблюдений требуется значительно больше данных. Поэтому алгоритмы, основанные на расстояниях, могут терять качество в пространствах высокой размерности.

Большая размерность также увеличивает потребление памяти и время вычислений, а нерелевантные признаки могут добавлять шум, из-за которого модели легче находить случайные закономерности.

**Вывод:** KNN и K-means особенно чувствительны к росту размерности, потому что их решения напрямую зависят от расстояний между объектами.

## 2. В чём разница между PCA и SVD?

**PCA (метод главных компонент)** — это метод снижения размерности, который ищет направления максимальной дисперсии данных.

**SVD (сингулярное разложение матрицы)** — это факторизация матрицы:

$$
X = U\\Sigma V^T
$$

где сингулярные значения находятся на диагонали матрицы $\\Sigma$.

Для PCA сначала центрируют матрицу признаков (из каждого признака вычитают его среднее значение), после чего SVD можно использовать для вычисления главных компонент.

Поэтому PCA и SVD — не одно и то же понятие:

- PCA — статистический метод поиска главных направлений и снижения размерности.
- SVD — общий метод линейной алгебры для факторизации матриц, который можно эффективно использовать для вычисления PCA.

**В этом проекте:** PCA является методом снижения размерности, а SVD — факторизацией матрицы, которая является одним из способов вычисления PCA.

## 3. В чём разница между NMF и SVD?

**NMF (неотрицательное матричное разложение)** раскладывает неотрицательную матрицу на неотрицательные множители:

$$
X \\approx WH,
$$

при этом

$$
W \\ge 0, \\qquad H \\ge 0.
$$

**SVD** не накладывает ограничения неотрицательности и раскладывает матрицу следующим образом:

$$
X = U\\Sigma V^T.
$$

Практическая разница особенно важна, когда матрица признаков изначально содержит только неотрицательные значения, например интенсивности пикселей изображения или TF-IDF. NMF сохраняет множители неотрицательными, поэтому полученные компоненты часто проще интерпретировать.

**Вывод:** SVD — общий метод разложения матрицы, а NMF — факторизация с ограничением неотрицательности, предназначенная для неотрицательных данных и часто дающая более интерпретируемые компоненты.

## 4. Опишите структуру алгоритма Locally Linear Embedding (LLE)

LLE — нелинейный метод снижения размерности. Его основная идея — сохранить локальные отношения между соседними объектами.

### Шаг 1. Найти соседей

Для каждого объекта выбираются его $k$ ближайших соседей в исходном пространстве высокой размерности.

### Шаг 2. Найти веса реконструкции

Для каждой точки находятся веса, с помощью которых точка как можно точнее восстанавливается через её соседей:

$$
x_i \\approx \\sum_{j \\in N(i)} w_{ij}x_j.
$$

Эти веса описывают локальную геометрию окрестности точки.

### Шаг 3. Найти низкоразмерное представление

Строятся низкоразмерные векторы $y_i$, использующие **те же веса реконструкции**:

$$
y_i \\approx \\sum_{j \\in N(i)} w_{ij}y_j.
$$

Таким образом, LLE старается сохранить локальные связи между соседями из исходного пространства в пространстве меньшей размерности.

**Вывод:** сначала LLE определяет, как каждая точка выражается через своих соседей, а затем ищет низкоразмерное представление, сохраняющее эти локальные зависимости.

## Итог

| Метод / понятие | Основная идея |
|---|---|
| Проклятие размерности | В пространствах высокой размерности расстояния становятся менее информативными, а для сохранения той же плотности требуется значительно больше данных |
| PCA | Ищет направления максимальной дисперсии |
| SVD | Разлагает матрицу в $U\\Sigma V^T$ и может использоваться для вычисления PCA |
| NMF | Разлагает неотрицательные данные на неотрицательные множители |
| LLE | Сохраняет локальные связи между соседними объектами при нелинейном снижении размерности |
